## Data and validation

```python
from analyze import analyze
import matplotlib.pyplot as plt
x = analyze()
assert len(x['counties']) == 47
print('Passed coverage, county joins, source-total and budget arithmetic checks.')
print('Known source differences:', x['quality'])
```

## Results
### The turnout denominator
IEBC declaration reproduced in Carter Center Appendix A, PDF pp. 50-52. All percentages use 22,120,458 declared registered voters, including diaspora. The non-voting residual does not reveal reasons.

```python
e=x['national_election']
labels=['Valid votes','Rejected ballots','Not recorded voting']
values=[100*e[k]/e['registered'] for k in ['valid_votes','rejected_ballots','not_recorded_voting']]
fig,ax=plt.subplots(figsize=(9,3.6))
ax.barh(labels,values,color=['#5b2a86','#b5433b','#8b8096'])
for i,v in enumerate(values): ax.text(v+1,i,f'{v:.2f}%',va='center')
ax.set_xlim(0,100);ax.set_xlabel('Share of declared registered electorate (%)');ax.set_ylabel('Recorded status')
ax.set_title('Kenya presidential election, 2022')
ax.spines[['top','right']].set_visible(False)
fig.tight_layout();plt.show()
```

### One numerator, two questions
Controller of Budget FY 2023/24, Table 2.3, PDF pp. 28-29. Development expenditure of KES 109,233.40 million divided by either the approved development budget or total expenditure. These are not service-quality measures.

```python
b=x['national_budget']
values=[b['development_pct'],b['development_share_of_spending_pct']]
fig,ax=plt.subplots(figsize=(9,3.6))
labels=['Of approved development budget','Of all county expenditure']
ax.barh(labels,values,color=['#5b2a86','#b5433b'])
for i,v in enumerate(values): ax.text(v+1,i,f'{v:.1f}%',va='center')
ax.set_xlim(0,100);ax.set_xlabel('Development expenditure / stated denominator (%)');ax.set_ylabel('Denominator')
ax.set_title('County development expenditure, FY 2023/24')
ax.spines[['top','right']].set_visible(False)
fig.tight_layout();plt.show()
```

PSC annual report FY 2023/24, Tables 5.2 and 5.10. Counts are from different Houses and may overlap. A processed question is not necessarily answered; a processed petition is not necessarily resolved.

## Takeaways
A civic-information product should show period, unit, denominator or stage, original source and interpretive limit beside every headline. This is a proposed reporting requirement, not a claim of tested civic impact.

Further questions need more evidence: project-completion records for services, bill-stage histories for legislation, and appropriate surveys for reasons behind turnout. Aggregate county data cannot establish individual voting behaviour.
